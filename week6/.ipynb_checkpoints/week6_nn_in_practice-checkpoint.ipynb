## 0. Setup

````python
import time
import copy
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
from torch.utils.data import DataLoader, TensorDataset

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score

torch.set_num_threads(4)    # use up to 4 CPU cores
DATA_FILE = Path("data/HIGGS_25k.csv.gz")
print(DATA_FILE, "found" if DATA_FILE.exists() else "NOT FOUND: copy it from week4/data")
````

````python
df = pd.read_csv(DATA_FILE, header=None, dtype=np.float32)
y = df[0].values                  # column 0 is the label: 1 = signal, 0 = background
X = df.drop(columns=0).values     # all 28 features

# 60 / 20 / 20 split, and scaling fitted on the training set only (Week 5)
X_rest, X_test, y_rest, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
X_train, X_val, y_train, y_val = train_test_split(X_rest, y_rest, test_size=0.25, random_state=42, stratify=y_rest)
scaler = StandardScaler().fit(X_train)
X_train, X_val, X_test = (scaler.transform(a).astype(np.float32) for a in (X_train, X_val, X_test))

# PyTorch versions of the arrays
train_ds = TensorDataset(torch.tensor(X_train), torch.tensor(y_train).view(-1, 1))
X_val_t = torch.tensor(X_val)
y_val_t = torch.tensor(y_val).view(-1, 1)

print(f"Training {len(X_train):,} | validation {len(X_val):,} | test {len(X_test):,}")
````

### The network and the training function

The same pieces as Week 4, with the options we'll explore as arguments. One small change: the network ends without a sigmoid, and the loss is `nn.BCEWithLogitsLoss`, which applies the sigmoid itself in a numerically safer way. (`torch.sigmoid` turns the output into a probability when you need one.)

`train_network` builds a network, trains it, and keeps the weights from the epoch with the **lowest validation loss**. It prints one line per network.

````python
loss_fn = nn.BCEWithLogitsLoss()


def make_network(width=256, depth=4, dropout=0.0):
    layers, n = [], 28
    for _ in range(depth):
        layers += [nn.Linear(n, width), nn.ReLU()]
        if dropout > 0:
            layers.append(nn.Dropout(dropout))
        n = width
    layers.append(nn.Linear(n, 1))
    return nn.Sequential(*layers)


def train_network(name, width=256, depth=4, dropout=0.0, lr=1e-3, weight_decay=0.0,
                  batch_size=256, epochs=60, patience=None, seed=0):
    """Build and train a network. Returns (model, history); the model keeps its best weights."""
    torch.manual_seed(seed)
    model = make_network(width, depth, dropout)
    torch.manual_seed(seed)
    loader = DataLoader(train_ds, batch_size=batch_size, shuffle=True)
    optimizer = torch.optim.AdamW(model.parameters(), lr=lr, weight_decay=weight_decay)

    hist = {"train_loss": [], "val_loss": [], "val_auc": []}
    best_loss, best_epoch, best_state = np.inf, 0, None
    t0 = time.time()
    for epoch in range(1, epochs + 1):
        model.train()                       # dropout on
        total = 0.0
        for xb, yb in loader:
            optimizer.zero_grad()
            loss = loss_fn(model(xb), yb)
            loss.backward()
            optimizer.step()
            total += loss.item() * len(xb)

        model.eval()                        # dropout off
        with torch.no_grad():
            out = model(X_val_t)
        hist["train_loss"].append(total / len(train_ds))
        hist["val_loss"].append(loss_fn(out, y_val_t).item())
        hist["val_auc"].append(roc_auc_score(y_val, out.numpy().ravel()))

        if hist["val_loss"][-1] < best_loss:
            best_loss, best_epoch = hist["val_loss"][-1], epoch
            best_state = copy.deepcopy(model.state_dict())
        if patience is not None and epoch - best_epoch >= patience:
            break                           # early stopping

    model.load_state_dict(best_state)
    hist.update(best_epoch=best_epoch, best_auc=hist["val_auc"][best_epoch - 1],
                epochs_run=epoch, seconds=time.time() - t0)
    print(f"{name:<28} best val loss {best_loss:.4f} at epoch {best_epoch:3d} | "
          f"val AUC there {hist['best_auc']:.4f} | {epoch} epochs, {hist['seconds']:.0f} s")
    return model, hist


def plot_val_loss(histories, title):
    plt.figure(figsize=(8, 4.5))
    for name, h in histories.items():
        plt.plot(range(1, len(h["val_loss"]) + 1), h["val_loss"], label=name)
    plt.xlabel("Epoch")
    plt.ylabel("Validation loss")
    plt.title(title)
    plt.legend()
    plt.grid(alpha=0.3)
    plt.show()
````

## 1. Overfitting

A network that's too big for the data: 4 hidden layers of 256 units, about 200,000 weights, trained for 60 epochs on 15,000 events.

````python
_, h_plain = train_network("No regularization")

epochs = range(1, len(h_plain["train_loss"]) + 1)
plt.figure(figsize=(8, 4.5))
plt.plot(epochs, h_plain["train_loss"], label="Training loss")
plt.plot(epochs, h_plain["val_loss"], label="Validation loss")
plt.axvline(h_plain["best_epoch"], color="gray", ls=":", label=f"Best epoch ({h_plain['best_epoch']})")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.title("A big network on a small dataset")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

The training loss keeps falling: the network gets better at the events it has seen. The validation loss falls for a few epochs, then **rises**: the network is memorizing the training events, and gets worse on new ones.

**Question:** if you only looked at the training loss, what would you conclude about epoch 60?

## 2. Regularization

Three standard ways to reduce overfitting, each judged on the validation set.

### Dropout

`nn.Dropout(p)` switches off a random fraction `p` of the hidden units on every batch during training. The network can't rely on any single unit, so it learns more robust patterns. `model.eval()` turns dropout off when making predictions.

````python
hists_reg = {"No regularization": h_plain}
for p in [0.2, 0.5]:
    _, hists_reg[f"Dropout {p}"] = train_network(f"Dropout {p}", dropout=p)

plot_val_loss(hists_reg, "Dropout")
````

### Weight decay

Weight decay shrinks every weight a little toward zero at each step, so the network keeps large weights only where they really help. It's the `weight_decay` argument of the optimizer.

With `AdamW`, each step shrinks the weights by a fraction `lr × weight_decay`. At our learning rate of 0.001 and only 60 epochs, small values do almost nothing, so we try large ones:

````python
for wd in [3.0, 10.0]:
    _, hists_reg[f"Weight decay {wd}"] = train_network(f"Weight decay {wd}", weight_decay=wd)

plot_val_loss({k: v for k, v in hists_reg.items() if "Dropout" not in k}, "Weight decay")
````

### Early stopping

Stop training when the validation loss hasn't improved for `patience` epochs, and keep the weights from the best epoch. `train_network` always keeps the best weights; `patience=5` also makes it stop early.

````python
_, h_early = train_network("Early stopping, patience 5", patience=5)
````

Same best model as the unregularized run, in a fraction of the time. Early stopping doesn't make the best model better; it finds it and stops. There's no reason not to use it.

### Side by side

````python
pd.DataFrame({
    name: {"best epoch": h["best_epoch"],
           "val AUC at best epoch": h["best_auc"],
           "val AUC at last epoch": h["val_auc"][-1]}
    for name, h in hists_reg.items()
}).T.round(4)
````

**Questions:**

1. Which method gives the best validation AUC?
2. Without regularization, how much do you lose by using the last epoch instead of the best one?
3. Weight decay 3 helps and 10 hurts. Why can too much regularization hurt?

## 3. Learning rate

The learning rate sets the size of each update step. Too small and training crawls; too large and the steps overshoot. It's usually the first setting to tune, on a log scale. We use a smaller network here (2 layers of 64 units) so overfitting doesn't hide the effect.

````python
hists_lr = {}
for lr in [1e-4, 1e-3, 1e-2, 1e-1]:
    _, hists_lr[f"lr = {lr}"] = train_network(f"Learning rate {lr}", width=64, depth=2, lr=lr, epochs=40)

plot_val_loss(hists_lr, "Learning rate (2 × 64 network)")
````

**Question:** which learning rate would you pick? What goes wrong at each end?

A common refinement is to start with a large learning rate and lower it during training, with a *scheduler* such as `torch.optim.lr_scheduler.CosineAnnealingLR`. It usually helps a little, not a lot.

## 4. Batch size

The batch size is how many events go into each update. Small batches give many noisy updates per epoch; large batches give a few smooth ones and run faster per epoch.

````python
hists_batch = {}
for bs in [32, 256, 2048]:
    _, hists_batch[f"batch {bs}"] = train_network(f"Batch size {bs}", width=64, depth=2, batch_size=bs, epochs=40)

plot_val_loss(hists_batch, "Batch size (40 epochs)")

pd.DataFrame({name: {"updates per epoch": int(np.ceil(len(train_ds) / int(name.split()[1]))),
                     "seconds per epoch": h["seconds"] / h["epochs_run"],
                     "val AUC at best": h["best_auc"]}
              for name, h in hists_batch.items()}).T.round(4)
````

**Question:** over 40 epochs, how many updates did the batch-2048 network get, compared with batch 32? Is the comparison fair?

### Putting it together: network against baseline

Take the best setting from Section 2 (dropout 0.5), add early stopping, and compare with logistic regression on the **test** set, once. This is the comparison every project should end with.

````python
final_model, _ = train_network("Final: dropout 0.5, early stop", dropout=0.5, epochs=100, patience=10)

baseline = LogisticRegression(max_iter=1000).fit(X_train, y_train)
final_model.eval()
with torch.no_grad():
    nn_scores = final_model(torch.tensor(X_test)).numpy().ravel()

print(f"Test AUC, logistic regression: {roc_auc_score(y_test, baseline.predict_proba(X_test)[:, 1]):.4f}")
print(f"Test AUC, neural network:      {roc_auc_score(y_test, nn_scores):.4f}")
````

## 5. Tools for your project

### 5a. Exoplanets: a 1D convolutional network for light curves

A light curve is a **sequence**, and a transit is a short dip that can appear anywhere in it. A fully connected network treats each point as a separate input, so it has to learn "a dip at position 10" and "a dip at position 40" separately.

A **convolutional** layer (`nn.Conv1d`) slides a small filter along the sequence, so the same dip detector is applied at every position. A max-pool at the end asks whether the detector fired anywhere.

To see the difference, we make a **simulated** dataset: 5,000 light curves of 64 points with noise, half with a dip of random position, width (3 to 8 points) and depth (0.6 to 1.4 times the noise).

````python
rng = np.random.default_rng(0)
N_CURVES, N_POINTS = 5000, 64

lc_y = rng.integers(0, 2, N_CURVES)
lc_X = rng.normal(0, 1, (N_CURVES, N_POINTS)).astype(np.float32)
for i in np.where(lc_y == 1)[0]:
    width = rng.integers(3, 9)
    start = rng.integers(0, N_POINTS - width)
    lc_X[i, start:start + width] -= rng.uniform(0.6, 1.4)

fig, axes = plt.subplots(2, 4, figsize=(15, 4.5), sharey=True)
for ax, i in zip(axes.ravel(), range(8)):
    ax.plot(lc_X[i], lw=1)
    ax.set_title("transit" if lc_y[i] else "no transit")
plt.tight_layout()
plt.show()
````

**Question:** can you tell by eye which curves have a transit?

The two models. `nn.Conv1d(1, 8, kernel_size=5)` learns 8 filters, each 5 points long. Conv1d expects inputs shaped (batch, channels, length), so we add a channel dimension with `unsqueeze(1)`.

````python
class TransitCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.layers = nn.Sequential(
            nn.Conv1d(1, 8, kernel_size=5, padding=2),    # 8 filters, 5 points wide
            nn.ReLU(),
            nn.MaxPool1d(2),                              # halve the length
            nn.Conv1d(8, 16, kernel_size=5, padding=2),   # 16 filters on top
            nn.ReLU(),
            nn.AdaptiveMaxPool1d(1),                      # strongest response anywhere
            nn.Flatten(),
            nn.Linear(16, 1),
        )

    def forward(self, x):
        return self.layers(x.unsqueeze(1))                # (batch, 64) -> (batch, 1, 64)


def train_lc(model, X, y, Xv, yv, epochs=40, seed=0):
    """A short training loop for the light curves; returns validation AUC per epoch."""
    torch.manual_seed(seed)
    loader = DataLoader(TensorDataset(torch.tensor(X), torch.tensor(y, dtype=torch.float32).view(-1, 1)),
                        batch_size=64, shuffle=True)
    opt = torch.optim.AdamW(model.parameters(), lr=1e-3)
    aucs = []
    for epoch in range(epochs):
        model.train()
        for xb, yb in loader:
            opt.zero_grad()
            loss_fn(model(xb), yb).backward()
            opt.step()
        model.eval()
        with torch.no_grad():
            aucs.append(roc_auc_score(yv, model(torch.tensor(Xv)).numpy().ravel()))
    return aucs


tr, va = slice(0, 3000), slice(3000, 4000)     # the last 1,000 curves are kept as a test set
lc_aucs = {}
for name, make in [("Fully connected (64-32)", lambda: nn.Sequential(
                        nn.Linear(N_POINTS, 64), nn.ReLU(), nn.Linear(64, 32), nn.ReLU(), nn.Linear(32, 1))),
                   ("1D CNN", TransitCNN)]:
    torch.manual_seed(0)
    m = make()
    lc_aucs[name] = train_lc(m, lc_X[tr], lc_y[tr], lc_X[va], lc_y[va])
    print(f"{name:<24} {sum(p.numel() for p in m.parameters()):5d} weights | "
          f"best val AUC {max(lc_aucs[name]):.3f} | last epoch {lc_aucs[name][-1]:.3f}")

plt.figure(figsize=(8, 4.5))
for name, a in lc_aucs.items():
    plt.plot(range(1, len(a) + 1), a, label=name)
plt.xlabel("Epoch")
plt.ylabel("Validation AUC")
plt.title("Finding dips in simulated light curves")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
````

The convolutional network does better with about a ninth of the weights, because its structure matches the problem. Building knowledge about the data into the network is often worth more than adding layers.

For real light curves, fold them at the period and bin them to a fixed length first (Week 5).

### 5b. Materials: ensembles and uncertainty

An **ensemble** trains several networks that differ only in their random seed, and averages their predictions:

1. The average is usually a little better than a single network, because their mistakes partly cancel.
2. Where the members **disagree**, the prediction is less certain. The spread between them is a simple uncertainty estimate. For a Materials regression (predicting a band gap, say), it's a ± on each prediction.

Here on the Higgs data, with 5 members:

````python
member_scores = []
for seed in range(5):
    m, _ = train_network(f"Member {seed}", width=128, depth=3, dropout=0.2, patience=10, seed=seed)
    m.eval()
    with torch.no_grad():
        member_scores.append(torch.sigmoid(m(X_val_t)).numpy().ravel())
member_scores = np.array(member_scores)          # shape: (5 members, validation events)

single_aucs = [roc_auc_score(y_val, s) for s in member_scores]
ensemble_mean = member_scores.mean(axis=0)
ensemble_spread = member_scores.std(axis=0)

print()
print(f"Single members: val AUC {min(single_aucs):.4f} to {max(single_aucs):.4f} (mean {np.mean(single_aucs):.4f})")
print(f"Ensemble mean:  val AUC {roc_auc_score(y_val, ensemble_mean):.4f}")
````

Does the spread mean anything? Split the validation events into five groups by how much the members disagree, and check how often the ensemble is right in each:

````python
groups = pd.qcut(ensemble_spread, 5, labels=["lowest", "low", "middle", "high", "highest"])
correct = (ensemble_mean > 0.5) == (y_val == 1)
pd.DataFrame({"spread": ensemble_spread, "correct": correct, "group": groups}) \
    .groupby("group", observed=True).agg(mean_spread=("spread", "mean"), accuracy=("correct", "mean")).round(3)
````

**Question:** how does the accuracy change as the members disagree more? How could you use that in a science analysis?

## 6. Summary

| Trick | In PyTorch | When |
|---|---|---|
| Watch for overfitting | plot training and validation loss | Always |
| Dropout | `nn.Dropout(p)` after hidden layers | Validation loss rises while training loss falls |
| Weight decay | `AdamW(..., weight_decay=...)` | Same; scan it on a log scale |
| Early stopping | keep the best epoch, stop after `patience` | Always |
| Learning rate | the optimizer's `lr` | Tune it first, on a log scale |
| Batch size | `DataLoader(..., batch_size=...)` | 64–512 is a good start |
| 1D convolution | `nn.Conv1d`, `nn.AdaptiveMaxPool1d` | Sequences: light curves, time series |
| Ensembles | same network, several seeds, average | Better scores and an uncertainty estimate |
