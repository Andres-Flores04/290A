````python
penguins.info()
````

Two penguins have no measurements at all, and some more are missing only `sex`. We'll plot the four measurements, so we drop only rows missing those (Tuesday's lesson: use `subset=`).

## 2. The anatomy of a figure

Every matplotlib plot today follows the same recipe:

```python
fig, ax = plt.subplots()      # a Figure (the canvas) holding one Axes (the plot)
ax.plot(...)                  # or ax.hist, ax.scatter, ...
ax.set_xlabel("...")          # ALWAYS label axes, with units
ax.set_ylabel("...")
ax.set_title("...")
ax.legend()                   # if there's more than one group
plt.show()
```

`fig, ax = plt.subplots(2, 3)` makes a grid of 2 × 3 plots; `ax` is then a 2D array of Axes.

**seaborn draws on the same Axes.** You hand it the whole DataFrame and name the columns:

```python
sns.scatterplot(data=penguins, x="bill_length_mm", y="bill_depth_mm", hue="species", ax=ax)
```

`hue=` colors by a column and adds the legend for you. You still label the axes with `ax.set_xlabel(...)`.

````python
# Colors for the three species, used for every plot today
COLORS = {"Adelie": "#2a78d6", "Chinstrap": "#eb6834", "Gentoo": "#1baf7a"}
SPECIES = list(COLORS)

# Readable axis labels for each measurement
LABELS = {"bill_length_mm": "Bill length (mm)", "bill_depth_mm": "Bill depth (mm)",
          "flipper_length_mm": "Flipper length (mm)", "body_mass_g": "Body mass (g)"}
````

## 3. Line plots: for ordered data

Line plots connect points in order, so they suit data with a natural order, usually **time**. Penguin measurements have no order, so we borrow Tuesday's Mauna Loa data.

````python
co2 = pd.read_csv(CO2_FILE, comment="#")
co2.columns = co2.columns.str.strip().str.replace(" ", "_")

fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(co2["decimal_date"], co2["average"], color="#d9622b", lw=1, label="Monthly mean")
ax.plot(co2["decimal_date"], co2["deseasonalized"], color="black", lw=1.5, label="Trend")
ax.set_xlabel("Year")
ax.set_ylabel("CO$_2$ (ppm)")
ax.set_title("Atmospheric CO$_2$ at Mauna Loa")
ax.legend()
ax.grid(alpha=0.3)
plt.show()
````

Try it: what does a line plot of penguin flipper length look like, in the order the rows happen to be in the file? Why is it meaningless?

````python
fig, ax = plt.subplots(figsize=(10, 3))
ax.plot(penguins["flipper_length_mm"].values, lw=0.8)
ax.set_xlabel("Row number in the file")
ax.set_ylabel("Flipper length (mm)")
ax.set_title("A line plot of unordered data: don't do this")
plt.show()
````

## 4. Histograms: how are values distributed?

A histogram splits the range into **bins** and counts how many values fall in each. The number of bins changes what you see.

````python
fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, nbins in zip(axes, [5, 20, 100]):
    ax.hist(penguins["flipper_length_mm"], bins=nbins, color="gray", edgecolor="white")
    ax.set_xlabel("Flipper length (mm)")
    ax.set_title(f"{nbins} bins")
axes[0].set_ylabel("Number of penguins")
plt.tight_layout()
plt.show()
````

**Question:** with 20 bins the distribution has two humps. What might cause that? (We'll find out in Section 6, it's the species clusters)

## 5. Scatter plots: how do two variables relate?

Each penguin becomes one point.

````python
fig, ax = plt.subplots(figsize=(6, 5))
ax.scatter(penguins["bill_length_mm"], penguins["bill_depth_mm"], s=15, color="gray")
ax.set_xlabel("Bill length (mm)")
ax.set_ylabel("Bill depth (mm)")
ax.set_title("All penguins")
plt.show()

r = penguins["bill_length_mm"].corr(penguins["bill_depth_mm"])
print(f"Correlation: {r:.2f}")
````

The correlation is **negative**: longer bills seem to go with shallower bills. Now color the points by species. First with matplotlib, which needs a loop over the species:

````python
fig, ax = plt.subplots(figsize=(6, 5))
for sp in SPECIES:
    sub = penguins[penguins["species"] == sp]
    ax.scatter(sub["bill_length_mm"], sub["bill_depth_mm"], s=15, color=COLORS[sp], label=sp)
ax.set_xlabel("Bill length (mm)")
ax.set_ylabel("Bill depth (mm)")
ax.set_title("Colored by species (matplotlib)")
ax.legend()
plt.show()
````

The same plot with seaborn. `hue="species"` replaces the loop and the legend:

````python
fig, ax = plt.subplots(figsize=(6, 5))
sns.scatterplot(data=penguins, x="bill_length_mm", y="bill_depth_mm",
                hue="species", palette=COLORS, ax=ax)
ax.set_xlabel("Bill length (mm)")
ax.set_ylabel("Bill depth (mm)")
ax.set_title("Colored by species (seaborn)")
plt.show()

# Correlation within each species
for sp in SPECIES:
    sub = penguins[penguins["species"] == sp]
    print(f"{sp:10s} {sub['bill_length_mm'].corr(sub['bill_depth_mm']):.2f}")
````

Within **every** species the correlation is **positive**. The negative overall trend appears only because the species sit in different places. This is **Simpson's paradox**: a trend in combined data can reverse inside each group.

The lesson for ML: always ask whether a hidden grouping is driving what you see.

## 6. Distributions separated by class

Next week we'll train classifiers to tell species apart. A good first question: **which measurement separates the classes best?** Overlaying one histogram per class shows it directly.

````python
fig, ax = plt.subplots(figsize=(7, 4))
sns.histplot(data=penguins, x="flipper_length_mm", hue="species", palette=COLORS,
             bins=30, ax=ax)
ax.set_xlabel("Flipper length (mm)")
ax.set_ylabel("Number of penguins")
plt.show()
````

The two humps from Section 4 were species! Gentoo penguins have longer flippers.

Now all four measurements, one panel each. Two options make classes of different sizes comparable:

- `stat="density"` scales each histogram to area 1, and `common_norm=False` does that **separately for each species** (there are fewer Chinstraps than Adelies)
- `element="step", fill=False` draws outlines, so overlapping histograms stay readable

````python
fig, axes = plt.subplots(2, 2, figsize=(11, 7))
for ax, col in zip(axes.flat, MEASUREMENTS):
    sns.histplot(data=penguins, x=col, hue="species", palette=COLORS, bins=25,
                 stat="density", common_norm=False, element="step", fill=False, lw=2, ax=ax)
    ax.set_xlabel(LABELS[col])
# seaborn adds a legend to every panel; keep only the first
for ax in axes.flat[1:]:
    ax.get_legend().remove()
plt.tight_layout()
plt.show()
````

**Questions:**
- Which single measurement best separates **Gentoo** from the others? [ flipper length ]
- Which separates **Adelie** from **Chinstrap**? [bill length]
- Is there any species pair that no single measurement separates well? What would you do then? [ adele vs chinstrap]

**Another view: box plots.** Each box spans the middle half of the values, the line is the median, and the whiskers show the spread. They're compact when there are many classes.

````python
fig, axes = plt.subplots(1, 4, figsize=(16, 4))
for ax, col in zip(axes, MEASUREMENTS):
    sns.boxplot(data=penguins, x="species", y=col, hue="species", palette=COLORS, legend=False, ax=ax)
    ax.set_xlabel("")
    ax.set_ylabel(LABELS[col])
plt.tight_layout()
plt.show()
````

## 7. Correlation matrices: which variables move together?

`df.corr()` computes the correlation between every pair of columns. Correlation runs from −1 (perfect opposite trend) through 0 (no *linear* trend) to +1 (perfect same trend).

````python
corr = penguins[MEASUREMENTS].corr()
corr.round(2)
````

````python
fig, ax = plt.subplots(figsize=(6, 5))
sns.heatmap(corr.rename(index=LABELS, columns=LABELS), annot=True, fmt=".2f",
            cmap="RdBu_r", vmin=-1, vmax=1, square=True,
            cbar_kws={"label": "Correlation"}, ax=ax)
ax.set_title("All penguins")
plt.show()
````

- `annot=True, fmt=".2f"` writes each value in its square, to 2 decimal places.
- `vmin=-1, vmax=1` with a red-white-blue colormap means white is always zero, so the colors mean the same thing on every plot.
- `.rename(index=LABELS, columns=LABELS)` swaps the column names for readable labels.

**Question:** flipper length and body mass have a correlation of 0.87. If you were building a model, would you need both? [ similar info]

## 8. Pair plots: every pair at once

A **pair plot** is a grid with a scatter plot for every pair of measurements, and each measurement's distribution on the diagonal. It's the standard first look at a new dataset. With seaborn it's one line:

````python
g = sns.pairplot(penguins, vars=MEASUREMENTS, hue="species", palette=COLORS,
                 height=2.6, plot_kws={"s": 15})
plt.show()
````

`sns.pairplot` makes its own figure (a grid of Axes), so it doesn't take `ax=`. It returns that grid, here called `g`, which we'll use to save it in Section 9. The diagonal shows smoothed histograms by default; try adding `diag_kind="hist"`.

## 9. Saving figures

Save the plots you'll use in reports and videos as image files. Save **before** `plt.show()`; `bbox_inches="tight"` trims extra white space.

````python
Path("figures").mkdir(exist_ok=True)

fig, ax = plt.subplots(figsize=(6, 5))
sns.scatterplot(data=penguins, x="bill_length_mm", y="bill_depth_mm",
                hue="species", palette=COLORS, ax=ax)
ax.set_xlabel("Bill length (mm)")
ax.set_ylabel("Bill depth (mm)")
fig.savefig("figures/bill_scatter.png", dpi=150, bbox_inches="tight")
plt.show()
print("Saved figures/bill_scatter.png")
````
