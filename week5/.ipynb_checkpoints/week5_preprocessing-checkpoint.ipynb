## 0. Setup

````python
import time
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.pipeline import make_pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.inspection import permutation_importance
from sklearn.metrics import (roc_auc_score, accuracy_score, precision_score,
                             recall_score, confusion_matrix, average_precision_score)

for f in ["data/magic04.data", "data/co2_mm_mlo.csv"]:
    print(f, "found" if Path(f).exists() else "NOT FOUND")
````

````python
FEATURES = ["fLength", "fWidth", "fSize", "fConc", "fConc1",
            "fAsym", "fM3Long", "fM3Trans", "fAlpha", "fDist"]

df = pd.read_csv("data/magic04.data", names=FEATURES + ["class"])
df["label"] = (df["class"] == "g").astype(int)     # 1 = gamma (signal), 0 = hadron (background)

X = df[FEATURES].values
y = df["label"].values

# The same 60 / 20 / 20 split as Week 3
X_rest, X_test, y_rest, y_test = train_test_split(X, y, test_size=0.20, random_state=42, stratify=y)
X_train, X_val, y_train, y_val = train_test_split(X_rest, y_rest, test_size=0.25, random_state=42, stratify=y_rest)

print(f"Training {len(X_train):,} | validation {len(X_val):,} | test {len(X_test):,}")
print(f"Fraction gamma: {y.mean():.3f}")
````

## 1. Scaling

### Why scale?

Look at the ranges of the features:

````python
df[FEATURES].describe().loc[["mean", "std", "min", "max"]].round(2)
````

`fDist` runs up to about 500, while `fConc1` stays below 1. Many models treat a difference of 1 the same in every feature, so a feature with large numbers dominates just because of its **units**, not because it matters more.

- **Distance-based** models (k-nearest neighbors, clustering) measure distances between events. Without scaling, the distance is mostly `fDist`.
- **Gradient-trained** models (logistic regression, neural networks) learn slowly or badly when features have very different ranges.
- **Trees** split on one feature at a time, at a threshold. Scaling doesn't change which events fall on each side, so trees don't care.

### Two common scalers

| Scaler | What it does | Result |
|---|---|---|
| `StandardScaler` | subtract the mean, divide by the standard deviation | mean 0, standard deviation 1 |
| `MinMaxScaler` | subtract the minimum, divide by the range | between 0 and 1 |

Both change the axis, not the shape of the distribution:

````python
scaled_std = StandardScaler().fit(X_train).transform(X_train)
scaled_mm = MinMaxScaler().fit(X_train).transform(X_train)

i = FEATURES.index("fLength")
fig, axes = plt.subplots(1, 3, figsize=(14, 3.5))
for ax, data, title in zip(axes, [X_train[:, i], scaled_std[:, i], scaled_mm[:, i]],
                           ["Raw (mm)", "StandardScaler", "MinMaxScaler"]):
    ax.hist(data, bins=60)
    ax.set_title(f"fLength: {title}")
plt.tight_layout()
plt.show()
````

### Fit the scaler on the training set only

The scaler learns numbers from the data (the mean and standard deviation, or the min and max). Like a model, it must learn them from the **training set only**, and then apply the same numbers to the validation and test sets. Fitting on all the data lets information from the test set leak into training.

A **pipeline** makes this automatic. `make_pipeline(StandardScaler(), LogisticRegression())` is one object: `.fit()` fits the scaler and then the model on the training data, and `.predict_proba()` applies both to new data.

### Does scaling matter? An experiment

Four models, each trained on raw, standardized and min-max-scaled features:

````python
models = {
    "k-nearest neighbors": lambda: KNeighborsClassifier(n_neighbors=25),
    "Logistic regression": lambda: LogisticRegression(max_iter=200),
    "Neural network": lambda: MLPClassifier(hidden_layer_sizes=(32, 32), max_iter=200, random_state=0),
    "Decision tree": lambda: DecisionTreeClassifier(max_depth=6, random_state=0),
}
scalers = {"Raw": None, "Standard": StandardScaler, "MinMax": MinMaxScaler}

rows = []
with warnings.catch_warnings():
    warnings.simplefilter("ignore")   # unscaled models warn that they didn't converge
    for m_name, make_model in models.items():
        row = {"model": m_name}
        for s_name, Scaler in scalers.items():
            model = make_model() if Scaler is None else make_pipeline(Scaler(), make_model())
            model.fit(X_train, y_train)
            row[s_name] = roc_auc_score(y_val, model.predict_proba(X_val)[:, 1])
        rows.append(row)

scaling_table = pd.DataFrame(rows).set_index("model").round(3)
scaling_table
````

Logistic regression gets the same AUC either way. But look at how hard it worked: `n_iter_` is the number of optimizer steps it took.

````python
with warnings.catch_warnings():
    warnings.simplefilter("ignore")
    raw = LogisticRegression(max_iter=200).fit(X_train, y_train)
scaled = make_pipeline(StandardScaler(), LogisticRegression(max_iter=200)).fit(X_train, y_train)

print("Steps, raw features:   ", raw.n_iter_[0], "(stopped at the limit of 200 without converging)")
print("Steps, scaled features:", scaled[-1].n_iter_[0])
````

**Questions:**

1. Which models gain most from scaling? Why those?
2. Why does the decision tree give the same answer in all three columns?
3. Standard vs. MinMax: does the choice matter much here?

### Outliers and skewed features

MinMax scaling uses the single largest and smallest values, so one extreme outlier squeezes everything else into a narrow band. `StandardScaler` is less sensitive, but heavy tails still hurt. For a feature with a long tail, a **log transform** before scaling often helps:

````python
fig, axes = plt.subplots(1, 2, figsize=(11, 3.5))
axes[0].hist(df["fLength"], bins=60)
axes[0].set_title("fLength")
axes[1].hist(np.log10(df["fLength"]), bins=60)
axes[1].set_title("log10(fLength)")
plt.tight_layout()
plt.show()
````

`fSize` in this dataset is already a log of the total brightness, for this reason. A log only works for positive values.

**Rule of thumb:** standardize by default; log-transform features that span several orders of magnitude; trees don't need either.

## 2. Class imbalance

In science, the interesting class is often **rare**: a few transiting planets among thousands of stars, a few Higgs events among billions of collisions, a few extreme-weather days in a year.

MAGIC is 65% gamma, which is fairly balanced. To see what imbalance does, we make a **rare-signal** version by keeping only 1 gamma in 20, in every set. The test set keeps the same ratio, because that's what the model would face in practice.

````python
rng = np.random.default_rng(0)

def make_rare(X, y, keep=0.05):
    """Keep all hadrons and a random 5% of gammas."""
    keep_mask = (y == 0) | (rng.random(len(y)) < keep)
    return X[keep_mask], y[keep_mask]

Xr_train, yr_train = make_rare(X_train, y_train)
Xr_val, yr_val = make_rare(X_val, y_val)
Xr_test, yr_test = make_rare(X_test, y_test)

for name, yy in [("training", yr_train), ("validation", yr_val), ("test", yr_test)]:
    print(f"{name:>10}: {len(yy):5,} events, {yy.sum():4d} gamma ({yy.mean():.1%})")
````

### Accuracy is misleading

A "model" that calls **every** event a hadron is right most of the time:

````python
always_hadron = np.zeros_like(yr_test)
print(f"Accuracy of always guessing hadron: {accuracy_score(yr_test, always_hadron):.3f}")
print(f"Gammas it finds:                    {recall_score(yr_test, always_hadron)}")
````

High accuracy, and it finds no gammas at all. With imbalanced classes, look at:

| Metric | Question it answers |
|---|---|
| **Recall** (efficiency) | Of the real gammas, what fraction did we find? |
| **Precision** (purity) | Of the events we called gamma, what fraction really are? |
| **ROC AUC** | How well does the score rank gammas above hadrons? Not affected by the ratio. |
| **Average precision** | Area under the precision–recall curve. Sensitive to the ratio; a random guess scores the gamma fraction. |
| **Confusion matrix** | All four counts at once |

### Three ways to train

1. **Plain:** train on the imbalanced data as it is.
2. **Class weights:** `class_weight="balanced"` makes each gamma count about 19 times as much as each hadron in the loss, so both classes contribute equally.
3. **Oversampling:** copy the gamma events (drawing with replacement) until the training set is balanced. Only ever oversample the **training** set.

````python
# Oversampling: draw gammas with replacement until there are as many as hadrons
gam = np.where(yr_train == 1)[0]
had = np.where(yr_train == 0)[0]
extra = rng.choice(gam, size=len(had) - len(gam), replace=True)
idx = np.concatenate([had, gam, extra])
Xr_over, yr_over = Xr_train[idx], yr_train[idx]
print(f"Oversampled training set: {len(yr_over):,} events, {yr_over.mean():.0%} gamma")
````

````python
variants = {
    "Plain": (make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)), Xr_train, yr_train),
    "Class weights": (make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000, class_weight="balanced")), Xr_train, yr_train),
    "Oversampling": (make_pipeline(StandardScaler(), LogisticRegression(max_iter=1000)), Xr_over, yr_over),
}

rows, confusions = [], {}
for name, (model, Xtr, ytr) in variants.items():
    model.fit(Xtr, ytr)
    score = model.predict_proba(Xr_test)[:, 1]
    pred = (score > 0.5).astype(int)              # the default threshold
    confusions[name] = confusion_matrix(yr_test, pred)
    rows.append({"training": name,
                 "accuracy": accuracy_score(yr_test, pred),
                 "recall": recall_score(yr_test, pred),
                 "precision": precision_score(yr_test, pred, zero_division=0),
                 "ROC AUC": roc_auc_score(yr_test, score),
                 "avg precision": average_precision_score(yr_test, score)})

imbalance_table = pd.DataFrame(rows).set_index("training").round(3)
imbalance_table
````

````python
for name, cm in confusions.items():
    print(name)
    print(pd.DataFrame(cm, index=["true hadron", "true gamma"], columns=["called hadron", "called gamma"]))
    print()
````

Class weights and oversampling give practically the same result here. For logistic regression they are nearly the same thing: counting each gamma 19 times, or including it 19 times.

**Questions:**

1. How does the recall change with class weights? What does it cost?
2. Why does the ROC AUC hardly change?
3. Weights and oversampling mostly move the **threshold**: the score above which we call an event gamma. You could get a similar effect by training plainly and choosing a threshold below 0.5. When might that be the simpler choice?

**Rule of thumb:** with imbalanced classes, report recall, precision and AUC, not accuracy. Try `class_weight="balanced"` first: it's one argument, and most scikit-learn classifiers accept it. In PyTorch, `nn.BCEWithLogitsLoss(pos_weight=...)` does the same job.

## 3. Feature selection

More features isn't always better: redundant or useless features add noise, slow training, and make results harder to interpret. Two simple tools.

### Correlation filtering

If two features are almost perfectly correlated, they carry nearly the same information, and you can often drop one.

````python
corr = df[FEATURES].corr()

plt.figure(figsize=(8, 6.5))
sns.heatmap(corr, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation between features")
plt.show()
````

````python
# List the pairs with |correlation| above 0.8
pairs = []
for i, a in enumerate(FEATURES):
    for b in FEATURES[i + 1:]:
        if abs(corr.loc[a, b]) > 0.8:
            pairs.append((a, b, round(corr.loc[a, b], 3)))
pd.DataFrame(pairs, columns=["feature 1", "feature 2", "correlation"])
````

Correlation only measures **linear** relationships between **pairs** of features. Two features can be uncorrelated and still redundant, and a feature that looks useless alone can matter in combination. Treat it as a first look.

### Importance scores

A random forest reports how much each feature was used in its splits: `feature_importances_`. This is quick, but it is computed on the training data and tends to favor features with many distinct values.

**Permutation importance** is more direct: shuffle one feature's values in the **validation** set, so it no longer carries information, and measure how much the AUC drops. A big drop means the model relied on that feature.

````python
forest = RandomForestClassifier(n_estimators=200, min_samples_leaf=5, n_jobs=-1, random_state=0)
forest.fit(X_train, y_train)
print(f"Random forest, all 10 features: validation AUC {roc_auc_score(y_val, forest.predict_proba(X_val)[:, 1]):.3f}")

t0 = time.time()
perm = permutation_importance(forest, X_val, y_val, scoring="roc_auc", n_repeats=5, random_state=0, n_jobs=-1)
print(f"Permutation importance took {time.time() - t0:.0f} s")

importance = pd.DataFrame({
    "forest (impurity)": forest.feature_importances_,
    "permutation (AUC drop)": perm.importances_mean,
}, index=FEATURES).sort_values("permutation (AUC drop)", ascending=False)
importance.round(3)
````

````python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
importance["forest (impurity)"].plot.barh(ax=axes[0])
axes[0].set_title("Forest importance (training data)")
axes[0].invert_yaxis()
importance["permutation (AUC drop)"].plot.barh(ax=axes[1], color="C1")
axes[1].set_title("Permutation importance (validation AUC drop)")
axes[1].invert_yaxis()
plt.tight_layout()
plt.show()
````

### How many features do we need?

Keep the top *k* features by permutation importance, retrain, and check the validation AUC:

````python
ranked = list(importance.index)
ks, aucs = [], []
for k in range(1, len(ranked) + 1):
    cols = [FEATURES.index(f) for f in ranked[:k]]
    m = RandomForestClassifier(n_estimators=200, min_samples_leaf=5, n_jobs=-1, random_state=0)
    m.fit(X_train[:, cols], y_train)
    ks.append(k)
    aucs.append(roc_auc_score(y_val, m.predict_proba(X_val[:, cols])[:, 1]))
    print(f"top {k:2d} ({ranked[k - 1]:>8}): validation AUC {aucs[-1]:.3f}")

plt.plot(ks, aucs, "o-")
plt.xlabel("Number of features kept")
plt.ylabel("Validation AUC")
plt.grid(alpha=0.3)
plt.show()
````

**Questions:**

1. How many features do you need to get within 0.01 of the full AUC?
2. Do the two importance methods agree on the top three?
3. Choosing features by their validation score and then reporting that same validation score is a mild form of cheating. Which set should you use to report the final result?

## 4. Features for your project

The most useful features usually come from knowing the science. One short example per project.

### 4a. Weather: rolling averages and lags

Weather data is a **time series**. Two standard tools:

- A **rolling average** smooths over a window (7 days, 30 days, 12 months) to show the trend without day-to-day noise.
- A **lag** feature is the value some steps earlier. To predict tomorrow, today's value is usually the best single feature.

Here on the Mauna Loa CO₂ record from Week 2, which has a clear trend and a clear seasonal cycle.

````python
co2 = pd.read_csv("data/co2_mm_mlo.csv", comment="#")
co2["date"] = pd.to_datetime(dict(year=co2["year"], month=co2["month"], day=15))
co2 = co2.set_index("date")[["average"]]

co2["rolling_12"] = co2["average"].rolling(window=12).mean()        # mean of this month and the 11 before
co2["seasonal"] = co2["average"] - co2["rolling_12"]                # what the rolling mean removed
co2["lag_1"] = co2["average"].shift(1)                              # last month's value
co2["lag_12"] = co2["average"].shift(12)                            # same month last year

co2.tail()
````

````python
recent = co2.loc["2015":]
fig, axes = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
axes[0].plot(recent.index, recent["average"], label="Monthly")
axes[0].plot(recent.index, recent["rolling_12"], label="12-month rolling mean")
axes[0].set_ylabel("CO2 (ppm)")
axes[0].legend()
axes[1].plot(recent.index, recent["seasonal"], color="C2")
axes[1].axhline(0, color="gray", lw=0.5)
axes[1].set_ylabel("Monthly minus rolling (ppm)")
plt.tight_layout()
plt.show()
````

The first 11 rows of `rolling_12` and the first 12 of `lag_12` are `NaN`: there isn't enough history yet. Drop those rows with `.dropna()` before training.

**Don't look into the future.** `rolling(window=12)` uses the current value and the 11 *before* it. `rolling(window=12, center=True)` would use 6 values *after* it, which a forecast can't know. Likewise, a lag must be `shift(1)` or more, never `shift(-1)`, except to make the **target** ("next month's value").

**For ERA5 in xarray**, the same operations work along `valid_time`:

```python
T_7day = T_chi.rolling(valid_time=7).mean()     # 7-day rolling mean
T_yesterday = T_chi.shift(valid_time=1)         # lag of one day
```

### 4b. Exoplanets: phase-folding

A transiting planet dims its star by a tiny amount for a few hours, once per orbit. The dip is often **smaller than the noise** in a single measurement. But it repeats with the orbital period. If you **fold** the light curve, stacking every orbit on top of the others by computing the time since the last transit, the dips add up and the noise averages away.

To see the idea clearly, we use a **simulated** light curve with a planet we put in ourselves: a 0.05% dip lasting 3 hours every 3.7 days, under noise of 0.1% per point, measured every 10 minutes for 90 days.

````python
rng_lc = np.random.default_rng(1)

P_TRUE, T0, DURATION, DEPTH = 3.7, 1.0, 3 / 24, 5e-4     # days, days, days, fraction
t = np.arange(0, 90, 10 / 60 / 24)                        # 90 days, one point every 10 minutes
flux = 1 + rng_lc.normal(0, 1e-3, len(t))                 # noise: 0.1% per point
in_transit = np.abs(((t - T0 + P_TRUE / 2) % P_TRUE) - P_TRUE / 2) < DURATION / 2
flux[in_transit] -= DEPTH

plt.figure(figsize=(12, 3))
plt.plot(t, flux, ".", ms=1)
plt.xlabel("Time (days)")
plt.ylabel("Relative brightness")
plt.title("Simulated light curve: can you see the planet?")
plt.show()
````

````python
def fold(t, period, t0):
    """Phase in days relative to the transit, from -period/2 to +period/2."""
    return (t - t0 + period / 2) % period - period / 2

def binned(phase, flux, half_width=0.5, nbins=48):
    """Mean flux in equal phase bins, within half_width days of the transit."""
    edges = np.linspace(-half_width, half_width, nbins + 1)
    which = np.digitize(phase, edges) - 1
    centers = (edges[:-1] + edges[1:]) / 2
    means = np.array([flux[which == b].mean() for b in range(nbins)])
    return centers, means

fig, axes = plt.subplots(1, 2, figsize=(13, 4), sharey=True)
for ax, period, title in [(axes[0], P_TRUE, f"Folded at the true period, {P_TRUE} d"),
                          (axes[1], 3.1, "Folded at a wrong period, 3.1 d")]:
    ph = fold(t, period, T0)
    ax.plot(ph * 24, flux, ".", ms=1, alpha=0.2, label="points")
    c, m = binned(ph, flux)
    ax.plot(c * 24, m, "o-", color="C3", ms=4, label="binned mean (30 min)")
    ax.set_xlim(-12, 12)
    ax.set_ylim(1 - 1.5e-3, 1 + 1.5e-3)
    ax.set_xlabel("Hours from transit")
    ax.set_title(title)
axes[0].set_ylabel("Relative brightness")
axes[0].legend(loc="upper right")
plt.tight_layout()
plt.show()
````

At the right period, the dip appears clearly. At the wrong period, the transits land at different phases and wash out.

**Finding the period.** In real data you don't know the period. A *Box Least Squares* (BLS) search tries many periods and reports how well a box-shaped dip fits at each one. astropy has one built in:

````python
from astropy.timeseries import BoxLeastSquares

bls = BoxLeastSquares(t, flux)
periods = np.linspace(1, 10, 20000)
result = bls.power(periods, DURATION)

best = periods[np.argmax(result.power)]
print(f"Best period: {best:.3f} days (true value {P_TRUE})")

plt.figure(figsize=(10, 3))
plt.plot(periods, result.power, lw=0.7)
plt.xlabel("Trial period (days)")
plt.ylabel("BLS power")
plt.show()
````

**Question:** the periodogram has smaller peaks besides the true period. Where are they, and why?

**With real Kepler or TESS data**, lightkurve has the same tools:

```python
pg = lc.to_periodogram(method="bls", period=np.linspace(1, 10, 20000))
folded = lc.fold(period=pg.period_at_max_power, epoch_time=pg.transit_time_at_max_power)
```

Phase-folded light curves, binned to a fixed number of points, also make good inputs for a classifier: every star gets a vector of the same length.

### 4c. Materials: features from a chemical formula

A model can't take `"Fe2O3"` as input. **matminer** turns a composition into numbers, using tabulated properties of each element. The `magpie` preset computes statistics (minimum, maximum, mean, ...) of 22 element properties over the atoms in the formula:

````python
warnings.filterwarnings("ignore", message="IProgress not found")   # a harmless notice from a progress-bar library

from pymatgen.core import Composition
from matminer.featurizers.composition import ElementProperty

featurizer = ElementProperty.from_preset("magpie")

formulas = ["Fe2O3", "Al2O3", "NaCl", "GaAs", "SiO2"]
feats = pd.DataFrame([featurizer.featurize(Composition(f)) for f in formulas],
                     index=formulas, columns=featurizer.feature_labels())
print(feats.shape[1], "features per formula")
feats[["MagpieData mean AtomicWeight", "MagpieData mean Electronegativity",
       "MagpieData range Electronegativity", "MagpieData mean NValence"]].round(2)
````

Everything from this notebook applies here: these features have wildly different scales (mean atomic weights from 20 to 72 in this table, electronegativities near 2), many are strongly correlated with each other, and with 132 of them, feature selection matters.

## 5. Summary

| Step | Tool | Watch out for |
|---|---|---|
| Scaling | `StandardScaler`, `MinMaxScaler`, in a pipeline | Fit on training data only. Trees don't need it. |
| Skewed features | `np.log10` | Positive values only |
| Class imbalance | `class_weight="balanced"`, oversampling | Don't judge by accuracy. Oversample training only. |
| Correlation filter | `df.corr()` | Only linear, pairwise |
| Importance | `permutation_importance` on validation | Report the final result on the test set |
| Time series | `.rolling()`, `.shift()` | Never use future values as features |
| Light curves | fold at the period, BLS search | |
| Compositions | matminer `ElementProperty` | Many correlated features |
